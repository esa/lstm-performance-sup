## Compute the in-band radiance for different Temperatures

```python
# Each entry is a channel: (wavelength in µm, response in µm-1, normalised to integrate to 1)
isrf = read_isrf("isrf_norm.csv")
```

```python
# Several temperatures, all TIR channels
T = np.array([250.0, 300.0, 350.0])
L_band = {name: band_radiance(T, wl, resp)
          for name, (wl, resp) in isrf.items() if name.startswith("TIR")}
```

```python
#Display integrated spectral radiance results for different temperatures and channels
df = pd.DataFrame(L_band, index=pd.Index(T, name="T [K]"))
df.round(3)
```
